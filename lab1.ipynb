5. What is the difference between \d+ and \W+ in regular expressions?

Ans: \d+ → Matches one or more digits (0–9).
Example: "abc123" → matches 123

\W+ → Matches one or more non‑word characters (anything except letters, digits, and underscore).
Example: "hello!!!" → matches !!!

1. Create a Python list containing the given text corpus.

```python
corpus = [
    "<p>Data Science is AMAZING!!! Visit https://datascience.com</p>",
    "Python 3.12 is released in 2025.",
    "NLP is used in AI.",
    "Contact us at info@gmail.com"
]
```

2. Import the required libraries (re and BeautifulSoup).

```python
import re
from bs4 import BeautifulSoup
```

3. Convert all text to lowercase.

```python
for text in corpus:
    text = text.lower()
    print(text)
```

```python
lowercase_corpus = [text.lower() for text in corpus]

print(lowercase_corpus)
```

4. Remove HTML tags.

```python
html_removed = [
    BeautifulSoup(text, "html.parser").get_text()
    for text in lowercase_corpus
]

print(html_removed)
```

5. Remove URLs

```python
url_removed = [
    re.sub(r'https?://\S+|www\.\S+', '', text)
    for text in html_removed
]

print(url_removed)
```

6. Remove email addresses.

```python
email_removed = [
    re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b', '', text)
    for text in url_removed
]

print(email_removed)
```

7. Remove digits.

```python
digit_removed = [
    re.sub(r'\d+', '', text)
    for text in email_removed
]

print(digit_removed)
```

8. Remove punctuation and special characters.

```python
punct_removed = [
    re.sub(r'\W+', ' ', text)
    for text in digit_removed
]

print(punct_removed)
```

9. Remove extra white spaces.

```python
cleaned_corpus = [
    re.sub(r'\s+', ' ', text).strip()
    for text in punct_removed
]

print(cleaned_corpus)
```

10. Store the cleaned text in a new list named cleaned_corpus.

```python
print("The cleaned text has been stored in the list 'cleaned_corpus'.")
print(cleaned_corpus)
```

11. Display the original corpus and the cleaned corpus

```python
print("Original Corpus:\n")

for sentence in corpus:
    print(sentence)

print("\nCleaned Corpus:\n")

for sentence in cleaned_corpus:
    print(sentence)
```

12. Compare the text before and after cleaning.

```python
print("Comparison Before and After Cleaning\n")

for original, cleaned in zip(corpus, cleaned_corpus):
    print("Original :", original)
    print("Cleaned  :", cleaned)
    print("-" * 60)
```

Sample Dataset

```python
# Sample dataset
reviews = [
    "This Laptop is AMAZING!!! <div>Battery lasts 10 hours</div> Visit https://laptopstore.com",
    "Worst laptop ever… Paid 60000 rupees, stopped working in 3 months!!!",
    "Email me at user123@techhelp.com if you want proof.",
    "Laptop performance is awesome <br> Highly Recommend!!!",
    "Not satisfied :( #slowLaptop"
]

import pandas as pd
df = pd.DataFrame(reviews, columns=['Review'])
df
```

Import Libraries

```python
import re
import pandas as pd
from bs4 import BeautifulSoup
```

Clean the Reviews

```python
def clean_text(review):
    review = review.lower()
    review = BeautifulSoup(review, "html.parser").get_text()   # remove HTML tags
    review = re.sub(r"http\S+|www\S+|https\S+", '', review)    # remove URLs
    review = re.sub(r'\S+@\S+', '', review)                    # remove emails
    review = re.sub(r'\d+', '', review)                        # remove digits
    review = re.sub(r'\W+', ' ', review)                       # remove punctuation/special chars
    review = re.sub(r'\s+', ' ', review).strip()               # remove extra spaces
    return review
```

Compare Before and After

```python
comparison = df[['Review', 'Cleaned_Review']]
comparison
```

Report 

Text cleaning greatly improves the quality of customer reviews for NLP tasks. The original dataset contained HTML tags, URLs, email addresses, digits, emojis, and excessive punctuation, which added noise. This ensures that algorithms focus only on meaningful words, improving the accuracy of tasks like sentiment analysis and topic modeling. While cleaning removes expressive elements such as emojis or repeated punctuation, the trade‑off is worthwhile because the dataset becomes structured and machine‑readable. Overall, text cleaning enhances data quality and model performance.

Conclusion

Text cleaning transforms messy, unstructured laptop reviews into clean, consistent text. This preprocessing step is essential in NLP pipelines, as it reduces noise, improves model performance, and ensures that the dataset is ready for deeper analysis.
